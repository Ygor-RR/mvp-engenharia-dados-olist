```python
# Documentando o Catálogo de Dados via SQL (Unity Catalog)

# 1. Tabela Fato Itens Pedidos
spark.sql("""
    COMMENT ON TABLE workspace.gold.fato_itens_pedidos IS 
    'Tabela fato central granular por item de pedido. Reúne métricas financeiras, logísticas de entrega e nota de avaliação do cliente.'
""")

spark.sql("COMMENT ON COLUMN workspace.gold.fato_itens_pedidos.order_id IS 'Identificador único do pedido (origem: orders).'")
spark.sql("COMMENT ON COLUMN workspace.gold.fato_itens_pedidos.order_item_id IS 'Sequencial do item dentro do pedido (1, 2, 3...).'")
spark.sql("COMMENT ON COLUMN workspace.gold.fato_itens_pedidos.product_id IS 'Chave estrangeira para dim_produto.'")
spark.sql("COMMENT ON COLUMN workspace.gold.fato_itens_pedidos.seller_id IS 'Chave estrangeira para dim_vendedor.'")
spark.sql("COMMENT ON COLUMN workspace.gold.fato_itens_pedidos.customer_id IS 'Chave estrangeira para dim_cliente.'")
spark.sql("COMMENT ON COLUMN workspace.gold.fato_itens_pedidos.data_compra IS 'Data da realização da compra (chave para dim_tempo).'")
spark.sql("COMMENT ON COLUMN workspace.gold.fato_itens_pedidos.order_status IS 'Status operacional do pedido (ex: delivered, shipped, canceled).'")
spark.sql("COMMENT ON COLUMN workspace.gold.fato_itens_pedidos.valor_item IS 'Preço unitário do produto em reais. Domínio: valores decimais >= 0.00.'")
spark.sql("COMMENT ON COLUMN workspace.gold.fato_itens_pedidos.valor_frete IS 'Custo de frete do item em reais. Domínio: valores decimais >= 0.00.'")
spark.sql("COMMENT ON COLUMN workspace.gold.fato_itens_pedidos.valor_total_item IS 'Soma de valor_item + valor_frete em reais.'")
spark.sql("COMMENT ON COLUMN workspace.gold.fato_itens_pedidos.dias_entrega_real IS 'Tempo decorrido entre a compra e a entrega real ao cliente em dias.'")
spark.sql("COMMENT ON COLUMN workspace.gold.fato_itens_pedidos.dias_entrega_estimada IS 'Prazo previsto informado no momento da compra em dias.'")
spark.sql("COMMENT ON COLUMN workspace.gold.fato_itens_pedidos.flag_atraso_entrega IS 'Indicador binário de atraso: 1 se entrega real > estimada, 0 se entregue no prazo.'")
spark.sql("COMMENT ON COLUMN workspace.gold.fato_itens_pedidos.review_score IS 'Nota média de avaliação do pedido. Domínio: 1.0 a 5.0.'")

# 2. Dimensão Produto
spark.sql("""
    COMMENT ON TABLE workspace.gold.dim_produto IS 
    'Dimensão descritiva dos produtos com categorias traduzidas para inglês.'
""")
spark.sql("COMMENT ON COLUMN workspace.gold.dim_produto.product_id IS 'Chave primária do produto.'")
spark.sql("COMMENT ON COLUMN workspace.gold.dim_produto.categoria_produto IS 'Nome da categoria traduzida para inglês ou nao_informado se nulo.'")
spark.sql("COMMENT ON COLUMN workspace.gold.dim_produto.peso_gramas IS 'Peso do produto embalado em gramas.'")

# 3. Dimensão Cliente
spark.sql("""
    COMMENT ON TABLE workspace.gold.dim_cliente IS 
    'Dimensão descritiva de clientes com localização geográfica de entrega.'
""")
spark.sql("COMMENT ON COLUMN workspace.gold.dim_cliente.customer_id IS 'Chave primária do cliente por pedido.'")
spark.sql("COMMENT ON COLUMN workspace.gold.dim_cliente.customer_unique_id IS 'Identificador único persistente do cliente no histórico.'")
spark.sql("COMMENT ON COLUMN workspace.gold.dim_cliente.customer_city IS 'Cidade de residência do cliente.'")
spark.sql("COMMENT ON COLUMN workspace.gold.dim_cliente.customer_state IS 'Sigla da Unidade Federativa (UF) do cliente (2 caracteres).'")

# 4. Dimensão Vendedor
spark.sql("""
    COMMENT ON TABLE workspace.gold.dim_vendedor IS 
    'Dimensão de parceiros vendedores com localização de origem do despacho.'
""")
spark.sql("COMMENT ON COLUMN workspace.gold.dim_vendedor.seller_id IS 'Chave primária do vendedor.'")
spark.sql("COMMENT ON COLUMN workspace.gold.dim_vendedor.seller_city IS 'Cidade de despacho do vendedor.'")
spark.sql("COMMENT ON COLUMN workspace.gold.dim_vendedor.seller_state IS 'Sigla da Unidade Federativa (UF) do vendedor.'")

# 5. Dimensão Tempo
spark.sql("""
    COMMENT ON TABLE workspace.gold.dim_tempo IS 
    'Dimensão calendário baseada na data de compra para agregações temporais.'
""")
spark.sql("COMMENT ON COLUMN workspace.gold.dim_tempo.data IS 'Data no formato AAAA-MM-DD (chave primária).'")
spark.sql("COMMENT ON COLUMN workspace.gold.dim_tempo.ano IS 'Ano da compra.'")
spark.sql("COMMENT ON COLUMN workspace.gold.dim_tempo.mes IS 'Mês numérico da compra (1 a 12).'")

# 6. Tabela Fato Pagamentos
spark.sql("""
    COMMENT ON TABLE workspace.gold.fato_pagamentos IS 
    'Tabela fato de liquidação e meios de pagamento utilizados por transação.'
""")
spark.sql("COMMENT ON COLUMN workspace.gold.fato_pagamentos.payment_type IS 'Método de pagamento: credit_card, boleto, voucher, debit_card.'")
spark.sql("COMMENT ON COLUMN workspace.gold.fato_pagamentos.payment_installments IS 'Número de parcelas selecionadas pelo cliente.'")
spark.sql("COMMENT ON COLUMN workspace.gold.fato_pagamentos.valor_pagamento IS 'Valor total liquidado na forma de pagamento em reais.'")

print("Comentários do Catálogo de Dados registrados com sucesso!")
```
